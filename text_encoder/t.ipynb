```python
class S(nn.Module):
    def __init__(self):

            super().__init__()
            self.nb=20
            self.op=nn.Linear(2,3)

            self.op.apply(self.weights_init)





    def weights_init(self,m):
        if isinstance(m, nn.Linear):
            nn.init.uniform_(m.weight,1,1)
            if m.bias is not None:
                nn.init.constant_(m.bias, 0)


    def forward(self,x):
        return self.op(x)
```
